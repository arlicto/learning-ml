<h1>Finding outliers</h1>

step 1: Find Q1 and Q2



step 2: find upperfence and lower fence


```upper fence = q3 + 1.5(iqr)```<br>
```lower fence = q1 + 1.5(iqr)```

```suppose: [1, 1, 2, 3, 4, 5, 6, 6, 7, 8, 9, 9, 10, 10, 11, 13, 13, 14, 15, 101]```

```uppper fence = 11 - 1.5(7) = 21.5```<br>
```lower fence = 4 -1.5(7) = -6.5```


any number greater than 21.5 is outlier and any number smaller than 6.5 is an outlier

````python
import numpy as np
def finding_outlier(any_list):
    n = len(any_list)
    q1 = np.percentile(any_list, 25)
    q3 = np.percentile(any_list, 75)
    print(q1)
    
    iqr = q3 - q1

    upper_fence = q3 + 1.5 * (iqr)
    lower_fence = q1 - 1.5 * (iqr)
    
    list_of_outliers = []
    for i in any_list:
        if i < lower_fence:
            print(f"outlier found: {i} ")
            list_of_outliers.append(i)
        elif i > upper_fence:
            print(f"outlier found: {i} ")
            list_of_outliers.append(i)
        
        
            
    print(f"\nThe total list of outliers are:\n")
    for i in list_of_outliers:
        print(f"{i}\n")

a = [-7,1, 1, 2, 3, 4, 5, 6, 6, 7, 8, 9, 9, 10, 10, 11, 13, 13, 14, 15, 101]
print(a)

finding_outlier(a)
````

#  Z-Test
- **Purpose:** Compares a sample mean or proportion with a population value, or compares two proportions, when assumptions are met.
- **ML use:** Testing whether a conversion rate differs significantly from a baseline.
- **Example:** Comparing conversion rates of two website designs.

````python
from scipy.stats import norm
# population std and mean should be given. if not use t-testa
# the sample size should be greater than 30(here we use sample size of less than 30 just for ease)
````

````python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]
population_mean = 170
population_std = 3
sample_mean = np.mean(sample)
n = len(sample)
````

````python
z_score = (sample_mean - population_mean) / (population_std/ np.sqrt(n))
z_score
````

````python
p_value = 2 * (1 - norm.cdf(abs(z_score)))
p_value
````

````python
alpha = 0.05
if p_value<alpha:
    print("reject the null hypothesis")
else:
    print("accept the null hypothesis")
````

#  One-Sample T-Test
- **Purpose:** Compares a sample mean with a known or hypothesized value.
- **ML use:** Checking whether the average prediction error differs from a target value.
- **Example:** Testing whether the average error differs from 5.

````python
from scipy import stats
````

````python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]
sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof= 1)
n = len(sample)
mean_population = 170
````

````python
t_stats= (sample_mean- mean_population)/(sample_std/np.sqrt(n))
t_stats
````

````python
p_value = 2 * (1 -stats.t.cdf(abs(t_stats), df= n -1))
alpha = 0.05
p_value
````

````python
if p_value < alpha:
    print("rejec null hypo")
else:
    print("accept null")
    
````

# 4. Two-Sample T-Test
- **Purpose:** Compares the means of two independent groups.
- **ML use:** Comparing the performance of two models or two groups.
- **Example:** Checking whether Model A and Model B have significantly different average errors

````python
group_A = [85, 88, 90, 92 , 87, 85, 91, 86, 88 ]
group_b = [82, 84,80,83,81,79,78,85,84,83]
````

````python
t_stats, p_value = stats.ttest_ind(group_A, group_b, equal_var=False)
````

````python
t_stats
````

````python
alpha = 0.05
if p_value < alpha:
    print("reject null hypo")
else:
    print("accept the null hypo")
````

#  ANOVA (Analysis of Variance)
- **Purpose:** Compares the means of three or more groups.
- **ML use:** Comparing the performance of multiple models or different configurations.
- **Example:** Comparing the accuracy scores of three machine learning algorithms.

````python
from scipy.stats import f_oneway
````

````python
df = sns.load_dataset('titanic')
df= df[['age', 'pclass']].dropna()
````

````python
class_1 = df[df['pclass'] == 1]['age']
class_2 = df[df['pclass'] == 2]['age']
class_3 = df[df['pclass'] == 3]['age']
````

````python
f_stats, p_value = f_oneway(class_1, class_2, class_3)
````

````python
f_stats
````

````python
p_value
````

````python
alpha = 0.05
if p_value < alpha:
    print('there is significant diff in atleast one passenger class')
else:
    print('there is no significant diff in atleast one passenger class')
    
````

# Covariance

- **Purpose:** Measures how two numerical variables change together.
- **Positive covariance:** Both variables tend to increase or decrease together.
- **Negative covariance:** When one variable increases, the other tends to decrease.
- **Zero covariance:** No linear co-movement is indicated.
- **ML use:** Understanding relationships between features and identifying patterns in data.
- **Example:** Studying how hours studied and exam scores change together.
- **Remember:** Covariance shows the direction of a relationship, while correlation measures its strength and direction on a scale from -1 to +1.

#  Correlation

- **Purpose:** Measures the strength and direction of the relationship between two numerical variables.
- **Positive correlation:** When one variable increases, the other tends to increase.
- **Negative correlation:** When one variable increases, the other tends to decrease.
- **Zero correlation:** No linear relationship between the variables.
- **ML use:** Understanding relationships between features and identifying patterns in data.
- **Example:** Studying the relationship between hours studied and exam scores.
- **Correlation coefficient:** Ranges from -1 to +1.
  - **+1:** Perfect positive correlation.
  - **-1:** Perfect negative correlation.
